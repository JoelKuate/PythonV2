# Les listes en Python

Une liste est une collection **ordonnée** et **modifiable**. Elle peut contenir des valeurs de types différents, y compris d’autres listes.

Dans cette leçon, vous allez apprendre à créer, lire, modifier, parcourir, trier et copier des listes.

## 1. Créer des listes

On écrit les éléments entre crochets `[]`, séparés par des virgules. Une liste peut être vide ou contenir plusieurs types de valeurs.

```python
nombres = [10, 20, 30]
melange = ["Ada", 36, True, None]
liste_vide = []

print(nombres)
print(melange)
print(list("chat"))  # ['c', 'h', 'a', 't']
print([0] * 4)       # [0, 0, 0, 0]
```

## 2. Lire et modifier les éléments

Les indices commencent à `0`. Les indices négatifs comptent depuis la fin. Un indice inexistant déclenche une erreur `IndexError`.

```python
lettres = ["a", "b", "c", "d"]
print(lettres[0])   # a
print(lettres[-1])  # d
print(lettres[1])   # b

lettres[1] = "B"   # remplacer un élément
print(lettres)
```

## 3. Ajouter et supprimer

- `append(x)` ajoute un élément à la fin.
- `insert(i, x)` insère à un indice.
- `extend(iterable)` ajoute plusieurs éléments.
- `remove(x)` supprime la première valeur égale à `x`.
- `pop(i)` supprime et renvoie un élément ; sans indice, retire le dernier.
- `del liste[i]` supprime par indice ; `clear()` vide la liste.

```python
fruits = ["pomme", "poire"]
fruits.append("kiwi")
fruits.insert(1, "banane")
fruits.extend(["figue", "raisin"])
print(fruits)

fruits.remove("poire")
dernier = fruits.pop()
print("Élément retiré :", dernier)
print("Liste restante :", fruits)
```

## 4. Tranches (slicing)

La syntaxe `liste[début:fin:pas]` sélectionne des éléments. Le début est inclus, la fin est exclue. Une tranche crée une nouvelle liste.

```python
nombres = [0, 1, 2, 3, 4, 5]
print(nombres[1:4])  # [1, 2, 3]
print(nombres[:3])   # [0, 1, 2]
print(nombres[3:])   # [3, 4, 5]
print(nombres[::2])  # [0, 2, 4]
print(nombres[::-1]) # [5, 4, 3, 2, 1, 0]
```

## 5. Parcourir, rechercher et mesurer

Utilisez `for` pour parcourir les valeurs et `enumerate()` si vous avez aussi besoin de l’indice. `in` teste la présence d’une valeur. `len`, `min`, `max` et `sum` sont des fonctions intégrées utiles.

```python
animaux = ["chat", "chien", "lapin"]
for indice, animal in enumerate(animaux):
    print(indice, animal)

nombres = [4, 2, 4, 8]
print(4 in nombres)       # True
print(nombres.count(4))  # 2
print(nombres.index(8))  # 3
print(len(nombres), min(nombres), max(nombres), sum(nombres))
```

## 6. Compréhensions de liste

Une compréhension permet de construire une liste de façon concise : `[expression for élément in iterable if condition]`. La partie `if` est facultative.

```python
carres = [x * x for x in range(6)]
pairs = [x for x in range(10) if x % 2 == 0]
types = ["pair" if x % 2 == 0 else "impair" for x in range(5)]

print("Carrés :", carres)
print("Pairs :", pairs)
print("Types :", types)
```

## 7. Trier et inverser

`sort()` trie la liste **sur place** (et renvoie `None`). `sorted()` crée une nouvelle liste triée. On peut préciser `reverse=True` ou une fonction `key`.

```python
nombres = [5, 1, 4]
triee = sorted(nombres)
print("Original :", nombres)
print("Triée :", triee)

nombres.sort(reverse=True)
print("Tri décroissant sur place :", nombres)

mots = ["banane", "Abricot", "cerise"]
print(sorted(mots, key=str.lower))
```

## 8. Copier une liste : attention aux références

L’affectation `b = a` ne copie pas la liste : `a` et `b` désignent le même objet. `a.copy()`, `a[:]` et `list(a)` créent une copie superficielle. Avec des listes imbriquées, les éléments internes restent partagés.

```python
originale = [1, 2]
reference = originale
copie = originale.copy()

reference.append(3)
print("Originale après modification de la référence :", originale)
print("Copie indépendante :", copie)

# Une copie profonde copie aussi les objets imbriqués.
import copy
matrice = [[1], [2]]
matrice_copie = copy.deepcopy(matrice)
matrice_copie[0].append(9)
print("Originale imbriquée :", matrice)
print("Copie profonde :", matrice_copie)
```

## 9. Listes imbriquées

Une liste peut contenir d’autres listes, par exemple pour représenter une matrice. Pour créer plusieurs lignes indépendantes, utilisez une compréhension plutôt que la multiplication d’une liste imbriquée.

```python
matrice = [
    [1, 2, 3],
    [4, 5, 6],
]
print(matrice[1][2])  # 6

# Bonne façon de créer une grille de lignes indépendantes :
grille = [[0] * 3 for _ in range(2)]
grille[0][0] = 9
print(grille)  # [[9, 0, 0], [0, 0, 0]]
```

## 10. Erreurs fréquentes et bonnes habitudes

- `list.sort()` modifie la liste et ne renvoie pas la liste triée ; utilisez `sorted()` pour obtenir une nouvelle liste.
- `remove(x)` et `index(x)` déclenchent `ValueError` si `x` n’est pas présent.
- Évitez de supprimer des éléments de la liste pendant que vous la parcourez ; construisez plutôt une nouvelle liste filtrée.
- Une recherche avec `x in ma_liste` parcourt les éléments. Pour des recherches d’appartenance fréquentes, un ensemble (`set`) peut convenir.
- Pour ajouter ou retirer souvent au début, pensez à `collections.deque`.
