# NumPy Essentiel pour Débutant

Ce notebook regroupe les bases indispensables de **NumPy** :
1. **Création de tableaux (`ndarray`)**
2. **Inspection des dimensions & types**
3. **Indexation & Découpage (*slicing*)**
4. **Opérations vectorisées (sans boucle `for`)**
5. **Filtrage conditionnel (masques)**
6. **Statistiques & Agrégations (axes `0` et `1`)**
7. **Changement de forme (`reshape`)**

## 1. Importation & Création de Tableaux

```python
import numpy as np  # "np" est le petit nom que tout le monde donne à NumPy

# Un tableau 1D (vecteur) : comme une liste Python, mais plus rapide et plus puissant
a = np.array([1, 2, 3, 4])
print("Tableau 1D :", a)

# Une matrice 2D : une liste de listes -> [lignes, colonnes]
# Ici : 2 lignes et 3 colonnes
b = np.array([[1, 2, 3], [4, 5, 6]])
print("Matrice 2D :\n", b)
```

## 2. Inspecter un Tableau (`shape`, `ndim`, `dtype`)

```python
# shape -> la "taille" du tableau sous forme (lignes, colonnes)
print("Dimensions (shape) :", b.shape)  # (2, 3) -> 2 lignes, 3 colonnes

# ndim -> le nombre de dimensions (1D, 2D, 3D...)
print("Nombre d'axes (ndim) :", b.ndim)  # 2 dimensions

# dtype -> le type des nombres stockés dans le tableau
print("Type de données (dtype) :", b.dtype)  # int64 ou int32
```

## 3. Indexation et Découpage (*Slicing*)

Règle clé : `tableau[lignes, colonnes]`

```python
m = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# m[ligne, colonne] -> on accède à UN élément précis (les index commencent à 0)
print("Élément ligne 0, col 1 :", m[0, 1])

# m[1, :] -> ":" veut dire "toutes les colonnes" -> on récupère la ligne 1 entière
print("Ligne 1 entière :", m[1, :])

# m[:, 0] -> ":" veut dire "toutes les lignes" -> on récupère la colonne 0 entière
print("Colonne 0 entière :", m[:, 0])

# m[0:2, 1:3] -> slicing : lignes 0 à 1 (2 exclu) ET colonnes 1 à 2 (3 exclu)
print("Sous-matrice (lignes 0-1, cols 1-2) :\n", m[0:2, 1:3])
```

## 4. Calculs Vectorisés (sans boucle `for`)

```python
x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

# Avec NumPy, pas besoin de boucle "for" : l'opération s'applique
# automatiquement à CHAQUE élément du tableau en une seule fois.
print("x + 10 :", x + 10)   # ajoute 10 à chaque élément -> [11 12 13]
print("x * 2  :", x * 2)    # multiplie chaque élément par 2 -> [2 4 6]
print("x ** 2 :", x ** 2)   # met chaque élément au carré -> [1 4 9]

# Entre deux tableaux de même taille, l'opération se fait élément par élément
print("x + y  :", x + y)    # [1+10, 2+20, 3+30] -> [11 22 33]
print("x * y  :", x * y)    # [1*10, 2*20, 3*30] -> [10 40 90]
```

## 5. Filtrage Conditionnel (Masques Booléens)

```python
valeurs = np.array([12, 5, 18, 7, 25])

# Un masque booléen : on teste la condition sur CHAQUE élément
# -> on obtient un tableau de True/False de la même taille
masque = valeurs > 10
print("Masque booléen :", masque)  # [ True False  True False  True]

# En utilisant le masque comme "filtre", on ne garde que les éléments
# qui correspondent à un True (donc qui respectent la condition)
selection = valeurs[masque]
print("Valeurs filtrées (> 10) :", selection)  # [12 18 25]
```

## 6. Statistiques et Agrégations (`axis=0` vs `axis=1`)

- `axis=0` : calcul le long des colonnes (résultat par colonne)
- `axis=1` : calcul le long des lignes (résultat par ligne)

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Sans préciser d'axe, le calcul se fait sur TOUT le tableau
print("Somme globale :", arr.sum())      # 1+2+3+4+5+6 = 21
print("Moyenne globale :", arr.mean())   # 21 / 6 = 3.5
print("Écart-type :", arr.std())         # dispersion des valeurs
print("Min / Max :", arr.min(), "/", arr.max())

# axis=0 -> on "descend" les colonnes : un résultat par colonne
print("Somme par colonne (axis=0) :", arr.sum(axis=0))  # [1+4, 2+5, 3+6] = [5 7 9]

# axis=1 -> on "traverse" les lignes : un résultat par ligne
print("Somme par ligne (axis=1) :", arr.sum(axis=1))    # [1+2+3, 4+5+6] = [6 15]
```

## 7. Redimensionnement (`reshape`)

```python
# np.arange(6) -> crée directement un vecteur [0, 1, 2, 3, 4, 5]
v = np.arange(6)
print("Vecteur 1D initial :", v)

# reshape -> réorganise les MÊMES données dans une nouvelle forme
# Attention : 2 x 3 = 6, il faut garder le même nombre total d'éléments
mat = v.reshape((2, 3))
print("Matrice 2x3 :\n", mat)

# flatten -> l'opération inverse : on "aplatit" une matrice en un vecteur 1D
plat = mat.flatten()
print("Aplatissement (flatten) :", plat)
```

### 🧠 Les 3 idées à garder en tête

1. **Un tableau NumPy (`ndarray`) ressemble à une liste Python, mais :**
   - toutes les valeurs ont le **même type** (souvent des nombres),
   - les opérations (`+`, `*`, `**`, comparaisons...) s'appliquent **automatiquement à chaque élément**, sans écrire de boucle `for`.

2. **La règle d'indexation à retenir : `tableau[lignes, colonnes]`**
   - `:` veut dire *"tout prendre sur cet axe"* (ex : `m[:, 0]` = toute la colonne 0).
   - Les index commencent toujours à **0**.

3. **`axis=0` vs `axis=1` (le point qui bloque souvent les débutants) :**
   - `axis=0` → on calcule **en descendant les colonnes** → 1 résultat par colonne.
   - `axis=1` → on calcule **en traversant les lignes** → 1 résultat par ligne.
   - Astuce mnémotechnique : *axis=0 = lignes qui bougent (vertical), axis=1 = colonnes qui bougent (horizontal)*.
