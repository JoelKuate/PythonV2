# Guide Pratique Pandas : Manipulation des Lignes et Colonnes

Ce notebook présente les manipulations fondamentales sur un DataFrame Pandas :
1. **Trier les colonnes** (ordre alphabétique / custom)
2. **Sélectionner une colonne** (Series vs DataFrame)
3. **Éliminer une colonne** (`drop`)
4. **Sélectionner et exclure des lignes** (`loc`, `iloc`, filtrage booléen)
5. **Cas pratique récapitulatif**

## 0. Initialisation et Jeu de Données d'Exemple

```python
import pandas as pd

# Création d'un DataFrame de test non trié
data = {
    'Salaire': [45000, 62000, 38000, 54000, 71000],
    'Age': [29, 41, 23, 35, 52],
    'Nom': ['Alice', 'Bob', 'Charlie', 'Diane', 'Éric'],
    'Ville': ['Paris', 'Lyon', 'Marseille', 'Lille', 'Bordeaux'],
    'Departement': ['IT', 'Finance', 'RH', 'IT', 'Direction']
}

df = pd.DataFrame(data, index=['id_101', 'id_102', 'id_103', 'id_104', 'id_105'])
print("--- DataFrame Initial ---")
df
```

## 1. Trier les Colonnes

On peut réordonner les colonnes :
- Par ordre alphabétique via `.sort_index(axis=1)`
- Selon un ordre personnalisé en passant une liste

```python
# 1.1 Tri alphabétique croissant
df_tri_alpha = df.sort_index(axis=1)
print("Colonnes triées par ordre alphabétique croissant :")
df_tri_alpha
```

```python
# 1.2 Tri alphabétique décroissant
df_tri_desc = df.sort_index(axis=1, ascending=False)
print("Colonnes triées par ordre alphabétique décroissant :")
df_tri_desc
```

```python
# 1.3 Réorganisation manuelle selon un ordre précis
ordre_voulu = ['Nom', 'Age', 'Departement', 'Ville', 'Salaire']
df_reordonne = df[ordre_voulu]
print("Colonnes réordonnées manuellement :")
df_reordonne
```

## 2. Sélectionner une Colonne

- Simple crochet `df['Nom']` : renvoie un objet **`pd.Series`** (1D)
- Double crochet `df[['Nom']]` : renvoie un **`pd.DataFrame`** (2D)
- Sélection par attribut `df.Nom` (valide si le nom n'a pas d'espace ni de conflit avec les méthodes pandas)

```python
# Sélection simple -> Series
col_series = df['Nom']
print("Type :", type(col_series))
print(col_series)
```

```python
# Sélection double crochets -> DataFrame
col_df = df[['Nom']]
print("Type :", type(col_df))
col_df
```

## 3. Éliminer une Colonne

On utilise `.drop()` en spécifiant `columns=['...']` (recommandé) ou `axis=1`.

```python
# Supprimer une seule colonne (retourne un nouveau DataFrame)
df_sans_ville = df.drop(columns=['Ville'])
print("DataFrame sans la colonne 'Ville' :")
df_sans_ville
```

```python
# Supprimer plusieurs colonnes en une seule opération
df_light = df.drop(columns=['Ville', 'Departement'])
print("DataFrame sans 'Ville' ni 'Departement' :")
df_light
```

## 4. Sélectionner une Ligne et en Exclure une Autre

Il existe 3 approches selon le besoin :
1. **Par label** (`.loc`)
2. **Par position entière** (`.iloc`)
3. **Par condition booléenne** (masque)

```python
# 4.1 Sélection par label d'index (.loc)
# Sélectionner 'id_101' et 'id_103' (en ignorant 'id_102')
selection_loc = df.loc[['id_101', 'id_103']]
print("Lignes sélectionnées par index : ", selection_loc)

# Exclure explicitement une ligne avec drop
df_sans_102 = df.drop(index='id_102')
print("DataFrame sans la ligne 'id_102' :", df_sans_102)
```

```python
# 4.2 Sélection par position (.iloc)
# Prendre la première ligne (index 0) et la troisième (index 2), exclure la deuxième (index 1)
selection_pos = df.iloc[[0, 2]]
print("Lignes aux positions 0 et 2 :")
selection_pos
```

```python
# 4.3 Sélection / Exclusion par condition logique
# Sélectionner toutes les lignes SAUF celle où Nom == 'Bob'
df_sans_bob = df[df['Nom'] != 'Bob']
print("DataFrame sans Bob : ", df_sans_bob)

# Sélectionner les profils avec Salaire > 50000 ET Departement != 'Direction'
df_condition = df[(df['Salaire'] > 50000) & (df['Departement'] != 'Direction')]
print("Salariés > 50k hors Direction :", df_condition)
```

## 5. Synthèse & Enchaînement (Method Chaining)

Toutes ces étapes peuvent être chaînées en une seule transformation fluide et lisible :

```python
resultat = (
    df
    .drop(columns=['Ville'])             # 1. Éliminer une colonne
    .sort_index(axis=1)                 # 2. Trier les colonnes alphabétiquement
    .loc[lambda d: d['Nom'] != 'Bob']   # 3. Exclure la ligne 'Bob'
)

print("--- Résultat du Pipeline ---")
resultat
```
